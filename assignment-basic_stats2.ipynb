```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df =pd.read_csv("Patient_Data.csv" )
```

```python
df.info()
```

```python
df = df.drop(columns = ["ReceptionistID", "CheckInTime"])
df= df[ ["PatientID", "Department", "Doctor", "BillAmount"] ]
df
```

```python
df =df.drop_duplicates(subset=["PatientID"])
df = df.reset_index(drop=True)
df
```

```python
df.head( 3 )
```

```python
avg = df["BillAmount"].mean()
df["BillAmount"] = df["BillAmount"].fillna(avg)
df
```

```python
dept = df.groupby("Department")["BillAmount"].sum()
dept = dept.reset_index()
dept
```

```python
bill =pd.read_csv( "Billing_Data.csv")
m = pd.merge(df, bill, on="PatientID" )
m
```
