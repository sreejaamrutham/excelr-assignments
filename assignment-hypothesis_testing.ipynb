```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x= np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
print("Mean:", x.mean())
print("Std:", x.std(ddof=1))
```

```python
from scipy import stats
n = len( x )
mean_val = x.mean()
std_val= x.std(ddof= 1)
t_crit = stats.t.ppf( 0.995, df=n-1 )
se = std_val / np.sqrt( n )
me= t_crit * se
ci_t = (mean_val - me, mean_val + me)
print( "99% CI (t):", ci_t )
```

```python
from scipy import stats
pop_std =0.2
z_crit = stats.norm.ppf(0.995)
se_z = pop_std / np.sqrt( n )
me_z= z_crit * se_z
ci_z = (mean_val - me_z, mean_val + me_z)
print("99% CI (z):", ci_z)
```

We use the t-distribution because sample size is small and population standard deviation is unknown. If the population standard deviation is known, we use the Z-distribution instead.

```python
from scipy import stats
x_bar= 3050
mu = 1000 + 5 * 600
sig = 5 * 25
n_h = 25
z_stat = (x_bar - mu) / (sig / np.sqrt(n_h))
print("Z-statistic:", z_stat)
p_val = 1 - stats.norm.cdf(z_stat)
print("P-value:", p_val)
```

Null Hypothesis: weekly cost is less than or equal to Rs. 4,000. Alternative Hypothesis: cost is greater than Rs. 4,000. The Z-score is -38 and p-value is 1.0, so we fail to reject the null hypothesis.
