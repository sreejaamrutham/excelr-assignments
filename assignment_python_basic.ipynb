```python
def is_prime( n ):
    if n <= 1:
        return False
    for i in range( 2, n ):
        if n % i == 0:
            return False
    return True
```

```python
is_prime( 11 )
```

```python
num= 29
ans= is_prime(num )
print( ans)
```
